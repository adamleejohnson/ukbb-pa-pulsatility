### Convert all data to nifti 

```python
root_folder = '{deepcmr_data_root}'
MAIN_DATAPATH_UNZIPPED = root_folder + '/cmr_lvot_20212_unzipped'
nifti_folder = root_folder + '/cmr_lvot_20212_niftis'
```

```python
folders = sorted(glob.glob(os.path.join(MAIN_DATAPATH_UNZIPPED, '*')))
```

```python
for k, folder in enumerate(folders):

    if k < 5068: continue
        
    try:
        opt = base_options.JupyterOptions(dataroot=folder, 
                                                   username=None,
                                                   max_dataset_size=float("inf"))


        dataset = dicom_dataset.DICOMDataset(opt)  

        for idx, (nifti, label, label_area) in enumerate(dataset):

            PatientName = '_'.join(dataset.acquisitions[0].split('_')[0].strip('^Bio').split(' '))
            image_path = os.path.join(nifti_folder, '%s.nii.gz'%(PatientName))
            nifti.to_filename(image_path)
            print(k, 'out of ', len(folders), PatientName, nifti.shape, nifti.header.get_zooms())
            
    except:
        print('='*20)
        print('Error', k, folder)
        print('='*20)
```
